# Public-method integration map

Report-only: no model fitting, feature construction, cloud calls, or Kaggle action. Public leaderboard values are external evidence, not local validation.

```python
from pathlib import Path
import json, pandas as pd
import plotly.express as px
from IPython.display import display
ROOT = Path.cwd().resolve()
while not (ROOT/'configs/top_solution_integration.json').is_file():
    if ROOT.parent == ROOT: raise RuntimeError('project root not found')
    ROOT = ROOT.parent
cfg = json.loads((ROOT/'configs/top_solution_integration.json').read_text())
print('Current private AUC:', cfg['current_private_auc'])
print('Winner private AUC:', cfg['winning_private_auc'])
print('Gap:', cfg['gap_to_winner'])
```

## Publicly documented systems above the current submission

Long method names use a horizontal layout and the height scales with row count.

```python
rows = pd.DataFrame(cfg['public_method_evidence'])
rows['delta_vs_current'] = rows['private_auc'] - cfg['current_private_auc']
rows = rows.sort_values('private_auc', ascending=True)
display(rows[['name','private_auc','delta_vs_current']])
height = max(500, min(950, 42*len(rows)+180))
fig = px.bar(rows, y='name', x='private_auc', orientation='h', hover_data={'delta_vs_current':':.5f'}, title='Publicly documented systems above the current 0.91425 private AUC')
fig.add_vline(x=cfg['current_private_auc'], line_dash='dash', annotation_text='Current 0.91425', annotation_position='top left')
fig.update_layout(autosize=True, height=height, margin=dict(l=240,r=45,t=90,b=70), showlegend=False)
fig.update_xaxes(title='Private leaderboard AUC', automargin=True)
fig.update_yaxes(title='', automargin=True)
fig.show(renderer='plotly_mimetype')
```
