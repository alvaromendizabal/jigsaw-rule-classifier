# Public-method integration map

Report-only: no model fitting, feature construction, cloud calls, or Kaggle action. Public solution write-ups are used as method references, not score targets.

```python
from pathlib import Path
import json, pandas as pd
import plotly.express as px
from IPython.display import display
ROOT = Path.cwd().resolve()
while not (ROOT/'configs/top_solution_integration.json').is_file():
    if ROOT.parent == ROOT: raise RuntimeError('project root not found')
    ROOT = ROOT.parent
cfg = json.loads((ROOT/'configs/top_solution_integration.json').read_text())
print('Retained project private AUC:', cfg['current_private_auc'])
print('Public references are reviewed for mechanisms, not leaderboard comparison.')
```

## Public method references reviewed

The table and chart summarize source availability for independently reimplemented mechanisms. They intentionally omit external leaderboard scores.

```python
rows = pd.DataFrame(cfg['public_method_evidence'])
rows['writeup_available'] = rows['url'].notna().astype(int) if 'url' in rows else 0
rows['code_available'] = rows['public_code'].notna().astype(int) if 'public_code' in rows else 0
rows['reference_artifacts'] = rows['writeup_available'] + rows['code_available']
display(rows[['name','code_license','writeup_available','code_available']].fillna(''))
height = max(420, min(850, 42*len(rows)+160))
fig = px.bar(rows, y='name', x='reference_artifacts', orientation='h', title='Public method references reviewed for independent implementation')
fig.update_layout(autosize=True, height=height, margin=dict(l=240,r=45,t=90,b=70), showlegend=False)
fig.update_xaxes(title='Available public reference artifacts', dtick=1, range=[0, 2.2], automargin=True)
fig.update_yaxes(title='', automargin=True)
fig.show(renderer='plotly_mimetype')
```
